### Reading Data From Different Sources


```python
import pandas as pd
from io import StringIO
Data = '{"employee_name": "James", "email": "james@gmail.com", "job_profile": [{"title1":"Team Lead", "title2":"Sr. Developer"}]}'
df=pd.read_json(StringIO(Data))
```

```python
df
```

```python
df.to_json()
```

```python
df.to_json(orient='records')
#records is a list of dictionaries where each dictionary represents a row in the DataFrame.
#This format is often used for data interchange and is compatible with many APIs and data processing tools.
```
