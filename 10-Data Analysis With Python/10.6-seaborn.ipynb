#### Data Visualization With Seaborn
Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lesson, we will cover the basics of Seaborn, including creating various types of plots and customizing them.

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
#Basic plottings with seaborn
tips=sns.load_dataset('tips')
tips
tips.info()
```

```python
#Create a scatter plot
sns.scatterplot(x='total_bill', y='tip', data=tips, color='green')
plt.title('Scatter Plot of Total Bill vs Tips')
plt.show()
```

```python
#Line Plot
sns.lineplot(x='size', y='total_bill',data=tips)
plt.title('Line Plot of Total Bill by Size')
plt.show()
```

```python
#categorical plot

#bar plot
sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Bar Plot of Total Bill by Day')
plt.show()
```

```python
#box plot
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Box Plot of Total Bill by Day')  
plt.show()
```

```python
#violin plot
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title('Violin Plot of Total Bill by Day')
plt.show()
```

```python
#histograms
sns.histplot(tips['total_bill'], bins=10, kde=True)
plt.title('Histogram of Total Bill')
plt.show()
```

```python
#KDE plot
sns.kdeplot(tips['total_bill'], fill=True)
plt.title('KDE Plot of Total Bill')
plt.show()
```

```python
#Pair Plot shows the relationship between all the numerical variables in the dataset
sns.pairplot(tips)
```

```python
#Heatmap shows the correlation between the numerical variables in the dataset
corr = tips[['total_bill', 'tip', 'size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
```

```python
import pandas as pd
```

```python
df = pd.read_csv('sales_data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
#totla salses by product

plt.figure(figsize=(10,6))
sns.barplot(x='Product Category', y='Total Revenue', data=df, estimator=sum)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Total Sales by Product')
plt.show()
```

```python
#total sales by region
#totla salses by product

plt.figure(figsize=(10,6))
sns.barplot(x='Region', y='Total Revenue', data=df, estimator=sum)
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.title('Total Sales by Region')
plt.show()
```
